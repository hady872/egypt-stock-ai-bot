```python
import pandas as pd
```

```python
df = pd.read_csv("../data/SKPC.csv")
df
```

```python
df.dtypes
```

```python
df["Data"] = pd.to_datetime(df["Data"])
df.dtypes
```

```python
df
```

```python
df.set_index("Data", inplace=True)
df
```

```python
df
```

```python
df["Daily_Return"] = df["Close"].pct_change()
df
```

```python
df.isna()
```

```python
df["MA_2"] = df["Close"].rolling(window=2).mean()
df
```

Day 4: Visualization and moving Average

```python
import matplotlib.pyplot as plt
```

```python
df
```

```python
plt.figure()
plt.plot(df["Close"])
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df["Close"])
plt.title("SKPC Closing price")
plt.xlabel("Date")
plt.ylabel("Price")
plt.show()
```

```python
df["MA20"] = df["Close"].rolling(20).mean()
```

```python
df.tail()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df["Close"])
plt.plot(df["MA20"])
plt.title("SKPC Price with MA20")
plt.show()
```
