```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df["Close"])
plt.title("SKPC Closing price")
plt.xlabel("Date")
plt.ylabel("Price")
plt.show()
```

```python
df["MA20"] = df["Close"].rolling(20).mean()
```

```python
df.tail()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df["Close"])
plt.plot(df["MA20"])
plt.title("SKPC Price with MA20")
plt.show()
```

```python
df[["Close", "MA20"]]
```

```python
df.tail()
```

```python
df["Trend"] = df["Close"] > df["MA20"]
df
```

```python
df["Trend"] = df["Trend"].map({True: "Uptrend", False: "Downtrend"})
df
```

```python
df.tail()
```

```python
latest_trend = df["Trend"].iloc[-1]
latest_price = df["Close"].iloc[-1]
latest_ma = df["MA20"].iloc[-1]

print("latest price:", latest_price)
print("moving average:", latest_ma)
print("Current trend:", latest_trend)
```

```python
df["Signal"] = "Hold"
df.loc[df["Close"] > df["MA20"], "Signal"] = "Buy"
df.loc[df["Close"] < df["MA20"], "Signal"] = "Sell"
```

```python
df.tail()
```

```python
latest_signal = df["Signal"].iloc[-1]
print("trend signal:", latest_signal)
```

```python
df["Diff"] = df["Close"] - df["MA20"]
```

```python
df["Cross Signal"] = "No Signal"
df.loc[(df["Diff"] > 0) & (df["Diff"].shift(1) <= 0), "Cross Signal"] = "Sell"
```

```python
df.tail()
```

```python
latest_cross = df["Cross Signal"].iloc[-1]
print("latest Cross Signal:", latest_cross)
```
